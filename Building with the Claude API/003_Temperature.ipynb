```python
# Create an API Client
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-5-5"
```

```python
# Make a request
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

## Note: `temperature` was removed from the Messages API

Earlier versions of this lesson passed `temperature=0.0` to `client.messages.create()`.
That parameter no longer exists: the current API/SDK only accepts
`max_tokens`, `messages`, `model`, `system`, `output_config`, `thinking`, `tools`, and so on.
The sampling parameters `temperature`, `top_p`, and `top_k` were removed in API version `2026-10-21`
(and from the Python SDK in v1.0), so no SDK version supports them. Passing `temperature` raises
`TypeError: Messages.create() got an unexpected keyword argument 'temperature'`.

There is no drop-in replacement. The official guidance is to omit them and guide behavior with prompting.
Related controls that still exist:

- `output_config={"effort": "low" | "medium" | "high" | "xhigh" | "max"}` -- how many tokens the model spends
- `thinking={"type": "adaptive"}` or `{"type": "disabled"}` -- whether Claude reasons first
- `output_config={"format": {...}}` -- structured outputs, useful for consistent shapes

Note that `temperature=0` never guaranteed identical output, and the API has no `seed` parameter.
To explore variation, call the model several times and compare the results.
